# Create a simple calculator that takes two numbers and an operator (+, -, *, /). Perform the selected operation and display an appropriate message if the operator is invalid.

```python
user1 = int(input("enter your num1 "))
user2 = int(input("enter your num2 "))
userop = input("enter operator")
if userop == "+" :
    print(user1+user2)
elif userop == "-" :
    print(user1-user2)
elif userop == "*" :
    print(user1*user2)
elif userop == "/" :
    print(user1/user2)
else:
    print("enter valid number")
```
